# 식별연산자
* is, is not (object, data type)이 같은지 비교

```python
print(a)
print(b)
print(c)
```

```python
print(a==b) # a는 정수 3, b는 실수 3.0이라 값이 같음
print(a is b) # 데이터 값이 같은지도 확인하기 때문에 다름
```

```python
d = 3
```

```python
print(a is d) # a와 d 모두 int 3이기 때문에 True
```
