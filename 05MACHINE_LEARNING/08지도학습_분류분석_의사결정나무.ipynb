# 분류 분석의 대표적인 알고리즘 Tree 기반 모델
# 의사결정나무 (Decision Tree)
* 의사결정나무란 ? 데이터를 분할해서 조건문처럼 분기해 나가며 예측하는 트리기반 모델
    * 스무고개와 비슷하게 계속 질문을 던져서 분ㄹ
### 장점
* 직관적이고 해석이 쉬움
* feature scling(변수)이 필요 없음(MinMax, SandardScaler 사용 하지 않아도 가능)
* 범주형(Decision Tree Classifier), 수치형(Decision Tree Regressor) 자료 모두를 분석할 수 있음
### 단점
* 과적합(overfitting)이 잘 일어남
* 최적 트리를 찾기 어렵고 시간이 오래 걸릴 수 있음



0. 필요한 라이브러리 로딩
1. 타이타닉 데이터 로딩
2. 데이터 전처리(결측값 처리, 이상값 탐지)
3. EDA(탐색적 데이터 분석, 각 컬럼의 데이터 분포, 그래프로, 문자 -> 순자, 상관관계)
4. Feature Enginerring(변수 선택, 새로운 특성 생성, 카테고리 변수의 더미 변수화(one-hot encoding))
5. 최정 선택된 변수를 독립변수(X)와 종속변수(Y)로 분리
6. sklearn의 train_test_split 불러와서 train, test 세트로 분리
7. sklearn에서 DecisionTreeClassfier를 불러와서 모델 훈련
8. test 세트를 모델에 넣어서 예측값 생성(pred)
9. accuracy_score(), classfication_report() 이용해서 모델 성능 확인
