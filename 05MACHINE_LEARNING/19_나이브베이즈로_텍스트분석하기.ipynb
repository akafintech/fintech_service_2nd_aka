```python
test_df = pd.read_csv("./data/ratings_test.csv.txt", sep="\t")
test_df.head()
```

```python
test_df = test_df.dropna()
test_df.info()
```

```python
train_df = test_df.dropna()
train_df.info()
```

```python
train_df['document']
```

* 정규표현식을 이용한 텍스트 정리

```python
import re
```

```python
def text_clean(x):
    pattern = r'[가-힣0-9a-zA-Z]+'
    matches = re.findall(pattern, x)
    matches = " ".join(matches)
    return matches
```

```python
train_df['document'] = train_df['document'].apply(text_clean)
```

```python
train_df
```

```python
X = train_df['document']
y = train_df['label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.3, random_state=10)
```

* document에 있는 문자를 숫자로 변환
* 문자를 벡터화
* CountVectorizer()

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer() # 토크나이저가 공백을 기준으로 토큰화
```

```python
cv.fit(X_train)
X_train = cv.transform(X_train)
X_valid = cv.transform(X_valid)
```

* 독립변수가 문자로 되어 있는 데이터의 경우 MultinominalNB

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
mnb = MultinomialNB()
mnb.fit(X_train, y_train)
pred = mnb.predict(X_valid)
print(classification_report(y_valid, pred))
```

* test 데이터 불러와서 검증

```python
X_test = test_df['document']
y_test = test_df['label']
```

```python
X_test = cv.transform(X_test)
```

```python
test_pred = mnb.predict(X_test)
print(classification_report(y_test, test_pred))
```
