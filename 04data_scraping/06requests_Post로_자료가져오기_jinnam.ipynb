```python
import requests
import time
import pandas as pd
from bs4 import BeautifulSoup as bs
```

```python
url = "https://kind.krx.co.kr/corpgeneral/corpList.do"
payload = dict(method="searchCorpList", pageIndex=1, currentPageSize=15, orderMode=3, orderStat="D", searchType=13, fiscalYearEnd="all", location="all")
```

```python
r = requests.post(url, data=payload)
print(r.status_code)
soup = bs(r.content, 'lxml')
soup
```

```python
com_list = soup.select("tbody > tr")
```

```python
time.sleep(5)
print(com_list)
```

```python
for com in com_list:
    print(com.td.contents)
    
```

```python
import bs4
table = soup.select_one("table")
key_list = table.attrs['summary'].replace(" ","").split(",")

result = {}
for com in com_list:
    real_contents = [content for content in com.contents if isinstance(content,bs4.element.Tag)]
    for key,content in zip(key_list,real_contents):
        if key == "홈페이지" and len(content.contents) >2:
            a_tag = content.contents[1]
            result.setdefault(key,[]).append(a_tag.attrs["href"])
        else:
            result.setdefault(key,[]).append(content.text)

df = pd.DataFrame(result)
df
```
