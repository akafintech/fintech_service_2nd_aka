```python
# 최초 1에서 100까지 조회한 것만 보여주는 코드
import requests
url = "https://openapi.naver.com/v1/search/shop.json"
# dict 사용할 때 {"key" : value} or dict(key = value)
# doct 왜 쓰는지?
# 파라미터에 값이 필요하다고(필수 여부Y or N) 네이버 api 가이드라인에 있음

params = {"query" : "노트북",
          "display" : 100,
          "start" : 1
         }

Client_id = "TaVTw3HnrpVGJDDGj3ym"
Client_secret = "46NsuGC17a"

headers = {"X-Naver-Client-Id" : Client_id,
           "X-Naver-Client-Secret" : Client_secret
          }
          
response = requests.get(url = url,
                        params = params,
                        headers = headers
)
# requests.status_codes : 일종의 설명서. 여러개의 요청 반응이 있음
# response.status_code : response(요청 반응)를 확인하는 것
data = response.json()
data
```
