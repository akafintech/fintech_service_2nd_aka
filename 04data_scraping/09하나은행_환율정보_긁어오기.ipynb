```python
import pandas as pd
from selenium import webdriver
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from datetime import datetime
from datetime import timedelta
from sqlalchemy import create_engine
import pymysql
# pymysql.install_as_MySQLdb()

service = Service(executable_path=ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)
driver.get('https://www.kebhana.com/cms/rate/index.do?contentUrl=/cms/rate/wpfxd651_01i.do')
driver.implicitly_wait(10)
table = driver.find_element(By.CSS_SELECTOR,"table.tblBasic.leftNone")

start_date = datetime(1995, 1, 3)
end_date = datetime(2025, 4, 8)

date_list = []
current = start_date
while current <= end_date:
     if current.weekday() < 5:
        date_list.append(current.strftime("%Y-%m-%d"))  
        current += timedelta(days=1)
        
    #tmpInqStrDt
    #tmpInqStrDt
        serch_box = driver.find_element(By.ID, "tmpInqStrDt")
        serch_box.clear()
        serch_box.send_keys(current.strftime("%Y%m%d")) # 현재 날짜 sendkey
      
        #조회 버튼 클릭
        #HANA_CONTENTS_DIV > div.btnBoxCenter > a
        button = driver.find_element(By.CSS_SELECTOR, "a.btnDefault.bg")
        button.click()
        time.sleep(5)
        table = driver.find_element(By.CSS_SELECTOR,"table.tblBasic.leftNone")
        df = pd.read_html(table.get_attribute('outerHTML'))
        break



def dbconnect():
    engine = create_engine("mysql+pymysql://root:1234@localhost:3306/exchange_rate_db")
    conn = engine.connect()
    return conn


def to_book_db(keyword, df):
    """
    keyword와 df 를 입력받아 네이버 도서에서 검색하고 5페이지 결과를
    {검색어 명}_book_info 테이블을 mysql에 저장
    """
    # Database 쿼리창 오픈
    conn = dbconnect()
    df.to_sql(f'{keyword}exchange_rate_db', con=conn,  if_exists="append", index=False)
    conn.close()
```
