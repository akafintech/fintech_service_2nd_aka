```python
all_data2 = []
start = 1
while True:
    url = "https://openapi.naver.com/v1/search/book.json"
    headers = {"X-Naver-Client-Id": "TaVTw3HnrpVGJDDGj3ym","X-Naver-Client-Secret" : "46NsuGC17a"}
    params  = dict(query=keyword, display = 100, start = 1000, sort = "sim")
    r = requests.get(url, params=params, headers=headers)
    print(r.url)
    print(r.status_code)
    data = r.json()
    n_data = data['total']
    total_pages = (int(data['total'])//100)+1
    print("전체데이터 검색 수 :", n_data, "검색할 페이지수:", total_pages)
    result_df = json_unpack(data)
    all_data2.append(result_df)
    if start == 1000 :
        break
    elif start < 901 :
        start += 100
    elif start == 901:
        start += 99
    
        
```
