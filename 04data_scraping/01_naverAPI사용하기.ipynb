```python
#택배회사
import requests

# 택배 보낼 때 필요한 거
주소 url : url = "https://openapi.naver.com/v1/search/blog.json"
# 헤더
client_id = "TaVTw3HnrpVGJDDGj3ym"
client_secret = "46NsuGC17a"
요청 방식 : GET
import urllib.parse

# requests 방식
client_id = "TaVTw3HnrpVGJDDGj3ym"
client_secret = "46NsuGC17a"
url = "https://openapi.naver.com/v1/search/blog.json"


# 요청방식이 어디있는지 몰랐다 ->
# requests.get()이 GET방식으로 요청하는 것
# 1.url 넣기
# url넣는 거는 requests.get(url=url)
# params 넣기
# 추가적인 tip(utf-8)
# 검색어는 query에 담아서 주는데 한글은 안받음
# 네이버 서버 받는 놈이 한글 안받는대
# 한글 말고 뭘로 줘? -> utf-8로 인코딩해서 줘야함
# 한글을 utf-8로 인코딩하는 방법은? -> urllib.parse.quote() 사용
# import urllib.parse
# encText = urllib.parse.quote("핀테크")
encText = urllib.parse.quote("핀테크")
display_count = 10
start_idx = 1

# 요청할 정보들을 모아서 params에 담기
# 서버 만든놈이 정한거라 서버에 따라 다 다름 잘 읽어보고 시키는대로 담아서 주면 됨
# get 요청시 requests.get(url=url,params=params)로 요청하면 됨
params = {
    "query": encText,  # 검색어
    "display": display_count,  # 검색 결과 개수 최대 100개 기본은 10개
    "start": start_idx,  # 검색 시작 위치 기본 1이고 100까지
    "sort": "sim"  # 정렬 방식 (sim: 유사도순, date: 날짜순)
}

# headers 넣기
headers = {
    "X-Naver-Client-Id": client_id,  # 클라이언트 아이디
    "X-Naver-Client-Secret": client_secret  # 클라이언트 시크릿
}
response = requests.get(url=url,
                        params=params, 
                        headers=headers)  # GET 요청

# 응답 결과 확인
# 200 번대면 성공
# 400 번대면 클라이언트 에러
# 500 번대면 서버 에러
if response.status_code == 200:
    print("스테이터스 코드",response.status_code)
    data = response.json()  # JSON 응답을 파싱
    print(data)  # JSON 응답을 파싱한 결과
```

```python
def text_clean2(x):
    result = x.replace("<b>", "").replace("</b>", "")
    return result
```

```python
def text_clean(x):
    # HTML 태그 제거
    result = re.sub(r"</?[^>]+>", "", x)
    # 한글, 영문, 숫자 외의 모든 문자 제거 (공백으로)
    result = re.sub(r"[^가-힣a-zA-Z0-9]", ' ', result)
    result = result.replace("  ", " ").replace("  ", " ").replace("  ", " ")
    return result
```

```python
result = {}
for item in data['items']:
    for key, value in item.items():
        value = text_clean(value)
        result.setdefault(key, []).append(value)
result
```
