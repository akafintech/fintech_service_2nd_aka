* sqlalchemy의 create_engine을 이용해서 mysql 서버에 접속<br>
engine = create_engine("mysql+pymysql://<userid>:<password>>@ip주소:<post>/데이터베이스이름)

```python
# localhost = 127.0.0.1
engine = create_engine("mysql+pymysql://root:1234@localhost:3306/stock_info")
# engine.connect : create_engine에 있는 정보로 db 접속
conn = engine.connect()
```

```python
# 데이터프레임을 db에 저장하기
# 데이터프레임.to_sql("테이블명")
df.to_sql(f"stock_company_info_", con=conn, if_exists = 'replace', index=False )
conn.close()
```
