# 반복문 for, while
* 반복문은 동일한 작업을 여러 번 반복해야 할 경우 사용
* for, while 두 종류의 반복문이 있다
* for는 **반복 횟수가 정해져 있는 경우**에 사용
* while은 **반복 횟수가 정해져 있지 않은 경우**에 사용(무한반복이 필요할때)

## for 반복문 : 반복할 횟수가 정해져 있는 경우
* for 변수1, 변수2 in 반복할 개체/자료:
____  반복할 코드

```python
# 숫자 생성 함수 RANGE(시작, 끝, STEP)
for num in range(1,11):
    print(num, end=" ")
```

* 순서가 있는 자료형(문자열, 리스트, 튜플_ + 순서는 없지만 dict도 반복문에서 사용 가능)

```python
# 문자열
for char in "파이썬은 문자열을 반복문에 넣어 출력가능":
    print(char, end="")
```

```python
# 리스트
for l_item in [1,2,3,"사","오",[6,7,8],(8,9,10),{"key":"value"}]:
    print(l_item)
```

```python
# 튜플
for l_item in (1,2,3,"사","오",[6,7,8],(8,9,10),{"key":"value"}):
    print(l_item)
```

```python
d=dict(이름="홍길동", 나이=40, 주소="서울", 전화번호="010-1111-1234")
d
```

```python
for item in d:
    print(d.get(item,"key 없음"))
```

```python
for item in d.keys():
    print(d[item])
```

```python
for item in d.values():
    print(item)
```

```python
for item in d.items():
    print(item)
```

```python
for key, value in d.items():
    print(key, value)
```

* for 반복문을 이용해서 1부터 10까지 정수를 더한 합을 출력하세요
    * 1부터 10까지 숫자를 만들어 반복문에 넣기
    * 1부터 10까지 누적 덧셈 하기
    * 누적 덧셈을 한 결과를 출력하기

```python
hap = 0
for num in range(1,11):
#     print(num, end=" ")
    print(f"{hap}+{num}", end=" :")
    hap += num # hap = hap +num
    print(f"{hap}")
    
    
```
