# 메소드(method)
* 클래스 안에 정의하는 함수
* 클래스에 기능을 만들어 준다
* def 메소드명(self, 매개변수1, *매개변수2="", **kwargs): <br>
____self.매개변수 = 매개변수
* self는 인스턴스 역할을 한다

* 사칙연산이 가능한 FourCal 계산기 만들기

````python
class FourCal() :
    # 데이터를 입력받는 역할의 setdata method
    def setdata(self, num1, num2):
        self.num1 = num1
        self.num2 = num2
        print("self.num1",self.num1)
        print("self.num2",self.num2)
        
    # 덧셈
    def add(self):
        result = self.num1 + self.num2
        return result
    # 뺄셈
    def sub(self):
        result = self.num1 - self.num2
        return result
    # 곱셈
    def mul(self):
        result = self.num1 * self.num2
        return result
    # 나눗셈
    def div(self) :
        result = self.num1 / self.num2
        return result
````

````python
sam = FourCal()
````

````python
sam.setdata(3,5)
````

````python
sam.add()
````

````python
sam.sub()
````

# 생성자
* 클래스를 실행해서 인스턴스를 만들 때 초기값을 받도록 해주는 메소드
* 생성자(constructor)는 객체(인스턴스)가 생성될 때 자동으로 호출되는 메서드, 주로 초기값을 받을 때 사용
* 클래스 내에서 메서드명으로 \_\_init\_\_ 으로 사용 됨

* 생성자를 이용해 초기값을 입력받는 FourCal2 만들기

````python
class FourCal2():
    def __init__(self,num1, num2) :
        self.num1 = num1
        self.num2 = num2
    def add(self) :
        result = self.num1 + self.num2
        return result
    def sub(self) :
        result = self.num1 + self.num2
        return result
    def mul(self):
        result = self.num1 * self.num2
        return result
    def div(self):
        result = self.num1 / self.num2
        return result
````

````python
gildong = FourCal2(3,4)
````

````python
class FourCal3():
    def __init__(self) :
        self.num1 = 0
        self.num2 = 0
    def add(self) :
        result = self.num1 + self.num2
        return result
    def sub(self) :
        result = self.num1 + self.num2
        return result
    def mul(self):
        result = self.num1 * self.num2
        return result
    def div(self):
#         self.num1 = num1
#         self.num2 = num2
        result = self.num1 / self.num2
        return result
````

# 클래스의 상속
* 어떤 클래스를 만들 때 다른 클래스의 기능을 물려받는 것
* 기존 클래스에 기능을 추가할 때 주로 사용
* A 클래스를 상속 받아서 B 클래스를 생성한다고 할 때, B 클래스는 A클래스의 모든 기능을 사용할 수 있다

* FourCal2를 상속받아 제곱 기능을 추가한 MoreFourCal 만들기

````python
class MoreFourCal(FourCal2):
    def pow(self):
        result = self.num1 ** self.num2
        return result
````

# 메서드 오버라이딩(method overrideing)
* 부모 클래스에서 상속한 메서드를 자식 클래스에서 동일한 이름으로 재정의 하는것


````python
kim = FourCal2(4,0)
````

````python
kim.add()
````

* FourCal2는 num2에 0이 입력될 경우 div 메서드 실행시 0으로 나누게 되기 때문에 에러 발생
* 해결하기위해 SafeFourCal을 만들고 FourCal2 상속받은 후에 div 함수를 재정의 하여 에러를 없애기

````python
class SafeFourCal(FourCal2) :
    def div(self):
        if self.num2 == 0:
            return 0
        else :
            return self.num1 / self.num2
````

````python
kim = SafeFourCal(4, 0)
````

````python
kim.div()
````

# 메서드 오버로딩
* 한 클래스 안에서 동일 메서드의 기능을 다르게 재정의 하는것
* 동일한 이름의 메서드가 기능이 약간 변경된 채로 여러 개 존재
* def add(num1, num2)
* def add(num1, num2, num3)
* def add(num1, num2, num3, num4)

````python
class OverloadingEx():
    def add(self, x, y) :
        return x + y
    def add(self, x, y, z):
        return x + y + z
    def add(self, x, y, z, a):
        return x + y + z + a
````

````python
sam = OverloadingEx()
````

````python
sam.add(10, 20, 30, 40)
````

````python
class OverloadingEx2():
    @dispatch(int, int)
    def add(self, x, y):
        return x + y
    @dispatch(int, int, int)
    def add(self, x, y, z):
        return x + y + z
    @dispatch(int, int, int, int)
    def add(self, x, y, z, a):
        return x + y + z - a
    @dispatch(float, float, float, float)
    def add(self, x, y, z, a):
        return (x + y) - (z * a)
````

# 정적메서드 static method
* 인스턴스를 만들지 않고도 클래스의 케서드를 바로 사용할 수 있는 메서드
* 메서드 위에 @staticmethod 라는 데코레이터를 붙여서 만든다.
* staticmethod는 self를 받지 않으므로 인스턴스 속성, 인스턴스 메서드가 필요하지 않을 때 사용한다

````python
class Cal() :
    @staticmethod
    def add(a, b):
        print(a+b)
        
        @staticmethod
        def sub(a, b):
            print(a-b)
````

# 클래스 메서드 class method
* 메서드 위에 @classmethod를 붙여서 만든다.
* 클래스 메서드의 첫 번째 매개변수에는 cls를 지정해야 한다.
* 클래스 메서드도 staticmethod처럼 인스턴스 생성 없이 호출이 가능
* 클래스 메서드는 메서드 안에서 클래스 속성, 클래스 메서드에 접근해야 할 때 사용

````python
class Factory() :
    n_coffee_machine = 0 #클래스 속성
    
    def __init__(self):
        # 인스턴스가 생성될 때 클래스 속성에 1을 더하기
        Factory.n_coffee_machine += 1 
        
    @classmethod
    def print_n_coffee_machine(cls):
        print(f"{cls.n_coffee_machine}개가 제조되었습니다.")
            
````

````python
Factory.print_n_coffee_machine()
````

````python
sam = Factory()
````

````python
sally = Factory()
````

````python
bob = Factory()
````

# 추상 클래스 abc(abstract base class)
* abc 모듈이 필요
* 추상클래스는 추상클래스를 상속받아 클래스를 만들 때 반드시 만들어야 하는 메서드를 지정해주는 클래스
* 다른 언어에서는 인터페이스라고 부름
* 규칙을 부여해서 그 규칙에 따라 클래스를 만들도록 강제하는 것
* 추상클래스를 상속받은 클래스는 반드시 추상클래스에서 정의한 메소드를 overridding으로 구현해야 한다.
* 이력서 양식에 맞추어 내용적기 같은 느낌

````python
from abc import *
````

````python
# 추상 클래스
class CoffeeBase(metaclass=ABCMeta):
    @abstractmethod
    def bean_input(self):
        pass
    
    @abstractmethod
    def grind(self):
        pass
    
````

````python
# 추상 클래스는 인스턴스 생성 불가
sam = CoffeeBase()
````

````python
class CoffeeMachine(CoffeeBase):
    def bean_input(self):
        print('커피 원두 넣기')
````

````python
samc=CoffeeMachine()
````

````python
class CoffeeBase2(CoffeeBase):
    def bean_input(self):
        print('커피 원두 넣기')
    
    def grind(self):
        print('커피 원두 갈기')
````

````python
sam2 = CoffeeBase2()
````

````python
sam2.bean_input()
````

````python
sam2.grind()
````
