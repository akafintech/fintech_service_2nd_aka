# 리스트 표현식
* 리스트 안에 for 반복문과 if else 조건문을 한번에 적어서 리스트를 만드는 방법
* 여러줄의 코드를 한줄로 줄일 수 있다.
* for문과 if else문을 사용해서 처리할 때보다 속도가 빠르다.
* [실행할 명령 for 변수 in list/tuple]

```python
for i in range(1,11):
    if i % 2 == 0 :
        print(f"{i}는 짝수입니다.")
    else:
        print(f"{i}는 홀수입니다.")
```

* 1부터 10까지 들어있는 리스트 만들기

```python
l1 = []
for i in range(1, 11):
    l1.append(i)
l1
```

```python
[i for i in range(1, 11)]
```

# 리스트 표현식과 일반 반복문의 속도 차이

```python
import time
```

```python
start = time.time()
a = []
for i in range(1,200000000):
    a.append(i+5)
print(a[:10])

print("코드 실행 시간:", time.time() - start)
```

```python
start2 = time.time()
b = [i+5 for i in range(1,200000000)]
print(b[:10])
print("코드 실행 시간:", time.time() - start2)
```

```python
import numpy as np
```

```python
start3 = time.time()
c = np.arange(1,20000000)
c = c+5
print(c[:10])
print("코드 실행 시간:", time.time() - start3)
```

```python
%%time
a = []
for i in range(1,200000000):
    a.append(i+5)
print(a[:10])

print("코드 실행 시간:", time.time() - start)
```

* 1에서 10까지의 숫자중 짝수만 더해서 리스트 만들기

```python
l1 = []
for i in range(1, 11):
    if i % 2 == 0 :
        l1.append(i)
l1
```

```python
[i for i in range(1, 11) if i % 2 == 0]
```

# 파일 읽기 쓰기 open(), with open()
## 1) 파일에 문자열을 쓰거나 읽을때는 open 함수 사용
* 파일저장 <br>
변수명 = open("파일이름", 모드(w:쓰기, r:읽기) <br>
변수명.write("저장할 내용") <br>
변수명.close()
**close()를 안하면 계속 열려있음

```python
file = open("test.txt", "w")
file.write("파일 만들기 연습")
file.close()
```

* 파일열기 <br>
변수명1 = open("파일이름", 파일모드(r)) <br>
변수명2 = 변수명.read() <br>
변수1.close()

```python
file1 = open("test.txt", "r")
my_file = file1.read()
file1.close()
print(my_file)
```

## with open
* with open은 close()를 하지 않아도 자동으로 close() 처리를 해줌

```python
with open("test2.txt", 'w') as file3:
    file3.write("with open을 사용하면 파일을 자동으로 닫아주어 편리함")
```

```python
with open ("test2.txt", 'r') as file4 :
    data = file4.read()
print(data)
```
