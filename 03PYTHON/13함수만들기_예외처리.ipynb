* num_maker2 - 정방향, 역방향 리스트가 출력되는 함수
* 3개의 숫자를 입력받아 숫자 리스트를 만들고
* 리스트+3을 한 결과를 리스트로 반환
* 또 한번은 역방향으로 출력

```python
def num_maker2(start, end, step):
    list_result = list(range(start, end, step))
    reversed_list = list_result[::-1]
    return list_result, reversed_list
```

```python
result=num_maker2(3, 30, 3)
print(result)
```

```python
for i in result:
    for j in i :
        print(j, end=" ")
```

# 중간에 함수를 종료시키기 위한 용도의 return
* 반복문에서 break처럼 활용 가능
* 함수 내에서 return 실행되면 함수는 종료

* 1에서 100까지 숫자중에서 짝수만 더해서 출력하시오.
* 단, 숫자가 50 이상이 되면 멈추시오
* 더해지는 과정도 출력

```python
num = list(range(1,101))
def even(num) :
    hap = 0
    for i in num :
        if i < 50 :
            if i % 2 == 0 :
                hap += i
                print(hap, end = " ")
        else:
            break
    return hap
```

```python
even(list(range(1,101)))
```

* 점수를 입력받아 91-100이면 A학점, 81-90 B학점, 71-80 C학점, 61-70 D학점, 60이하 F학점을 출력하는 student_grade 함수를 만드세요
* 점수받기 input()
* student_grade 함수에 입력받은 값을 넣어줌
* student_grade 함수에서는 입력받은 값을 비교해서 학점출력

```python

def student_grade(score) :
        if 91 <= score <= 100:
            return "A학점"
        elif 81<= score <=90 :
            return "B학점"
        elif 71<= score <= 80 :
            return "C학점"
        elif 61<= score <= 70 :
            return "D학점"
        else:
            return "F학점"
        return score
        
        
```

```python
grade = int(input("학점을 입력하세요"))
result = student_grade(grade)
print(result)
```

* 위치인수와 가변인수 동시에 받기
* 위치인수와 가변인수를 동시에 받을 때는 반드시 위치인수가 먼저 와야함
* def muls(a, *args): OK
* def muls(*args, a) : X

```python
# 정수와 리스트를 입력받아 리스트의 요소에 입력받은 정수를 곱하는 함수
```

```python
def muls(a, *args):
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
l2 = [1,2,3,4,5]
```

```python
muls(6, *l2)
# muls(1,2,3,4,5,2)
# muls(*l2, 2)
```

* 가변인수(*args)가 앞으로 오면 a로 갈 자료가 없음

```python
def muls(*args,a):
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
muls(*l2,3)
```

```python
def personal_info6(idx, addr="비공개", *args, **kwargs):
    print(f"idx:{idx}")
    print(f"addr:{addr}")
    print(f"*args:{args}")
    print(f"**kwargs:{kwargs}")
```

* 1에서 10까지의 수 중에서 3의 배수만 문자열로 바꾸기

```python
def tostr(x):
    result=[]
    for i in x :
        if i % 3 == 0:
            result.append(str(i))
        else:
            result.append(i)
    return result
```

```python
tostr(list(range(1,11)))
```

```python
result_str = tostr(list(range(1,11)))
print(result_str)
```
