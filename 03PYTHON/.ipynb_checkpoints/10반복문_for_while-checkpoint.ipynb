# 반복문 for, while
* 반복문은 동일한 작업을 여러 번 반복해야 할 경우 사용
* for, while 두 종류의 반복문이 있다
* for는 **반복 횟수가 정해져 있는 경우**에 사용
* while은 **반복 횟수가 정해져 있지 않은 경우**에 사용(무한반복이 필요할때)

## for 반복문 : 반복할 횟수가 정해져 있는 경우
* for 변수1, 변수2 in 반복할 개체/자료:
____  반복할 코드

```python
# 숫자 생성 함수 RANGE(시작, 끝, STEP)
for num in range(1,11):
    print(num, end=" ")
```

* 순서가 있는 자료형(문자열, 리스트, 튜플_ + 순서는 없지만 dict도 반복문에서 사용 가능)

```python
# 문자열
for char in "파이썬은 문자열을 반복문에 넣어 출력가능":
    print(char, end="")
```

* for 반복문을 이용해서 1부터 10까지 정수를 더한 합을 출력하세요
    * 1부터 10까지 숫자를 만들어 반복문에 넣기
    * 1부터 10까지 누적 덧셈 하기
    * 누적 덧셈을 한 결과를 출력하기

```python
hap = 0
for num in range(1,11):
#     print(num, end=" ")
    print(f"{hap}+{num}", end=" :")
    hap += num # hap = hap +num
    print(f"{hap}")
    
    
```

```python
* 1부터 50까지 수 중에서 짝수인 수의 합을 구하시오
* 1-50 수를 만들기 (range(1-51))
* 홀수, 짝수 구분 if, %2 == 0
* 짝수일 때 누적 합, 결과 저장할 변수 hap
* 최종결과 출력
```

```python
hap=0
for num in range(1,51):
#     print(num, end="")
    if num % 2 == 0 else print(f"{hap}+{num}" ,end=" :")
#     hap += num(2n+1)
#     print(f"{hap}")
```

```python
hqp = 0
for num in range(1,51) :
    print(num, end="")
    if num % 2 == 0:
        print(f"hap:{hqp}, num: {num}")
        hap += num
print(hap)
```

* 1부터 100까지 수를 만들고 그중에서 30-70까지만 더해서 출력
* 과정 출력
* 두가지 방법을 이용해서 코드 작성 - if조건문, 슬라이싱

```python
hap = 0
for num in range (1,101) :
#     print(num, end=" ")
    if 70 >= num >= 30 :
        print(f"hap : {hap}, num : {num}")
        hap += num
print(hap)
```

```python
hap = 0
for num in range(1, 101):
#     print (num, end=" ")
    if 30 <= num <= 70 :
        print(f"hap:{hap}, num:{num}")
        hap += num
        print(hap)
        
```

```python
hap=0
for num in range(1,101)[29:70:2]:
#     print(num, end=" ")
    print(f"hap:{hap}, num{num}")
    hap+num
print(hap)
```

* 100에서 1까지 더하는 반복문 만들기
* range에서 역순으로 숫자 생성
* range(1, 101)을 뒤집는 방법 2가지
* 최종 결과만 출력

```python
hap = 0
for num in range(100,0, -1):
#     print(num, end=" ")
    hap += num
print(hap)
```

* 슬라이싱을 이용해서 순서 뒤집기

```python
hap = 0
for num in range(1,101)[::-1]:
    print(num, end=" ")
    hap += num
print(hap)
```

* reversed() 리스트를 역순으로 뒤집는 함수

```python
hap = 0
for num in reversed(range(1,101)):
    print(num, end=" ")
    hap += num
print(hap)
```

## enumerate() 반복문에서 자료의 인덱스를 만들어주는 함수

```python
s = "enumerate 함수는 반복문에서 idnex를 만들어 출력해주는 함수입니다."
```

```python
import time
```

```python
for idx, i in enumerate(s):
    print(f"현재 반복 횟수: {idx+1}/{len(s)}", end="\r")
    time.sleep(1)
#     print(idx, i)
```

## continue, break
* continue : 코드 실행을 건너뛰고 반복 계속 진행
* break : 특정 조건이 되었을 때 정지

```python
for idx, i in enumerate(s):
#     print(f"현재 반복 횟수: {idx+1}/{len(s)}", end="\r")
    if idx % 2 == 0 :
        continue
    else:
        print(idx, i)
    time.sleep(1)
```

```python
for idx, i in enumerate(s):
#     print(f"현재 반복 횟수: {idx+1}/{len(s)}", end="\r")
    if idx % 2 == 0 :
        continue
    else:
        print(idx, i)
        if idx >= 19:
            break
            
    time.sleep(1)
```

# while 반복할 횟수가 정해지지 않은 경우
* 무한 반복 가능
* 무한 반복이 되지 않도록 조건식을 잘 짜주는 것이 중요

변수 선언 및 초기화<br>
while 조건식:<br>
____ 반복할 코드<br>
____ 변수 증가/감소 코드<br>
____ 반복문이 종료될 조건 + break<br>

while문으로 1-10까지 출력하기


```python
i = 1
while i <= 10:
    print(i, end=" ")
    i += 1
```

* 20에서 1까지 숫자를 더한값 출력

```python
i=20
hap = 0
while i > 0:
    print(f"hap:{hap}, i:{i}")
    hap += i
    i -= 1
print(hap)
```

* 1-50까지의 수 중에서 3의 배수이거나 짝수인 수를 찾아 더한 값을 출력하세요
* 1-50까지 숫자 생성 후 찍어보기
* if문으로 3의배수 %3 == 0, or 짝수 %2==0 인 경우
* hap에 더함

```python
i= 1
hap = 0
while i <= 50:
    print(f"hap:{hap}, i:{i}")
    if i % 3 == 0 or i % 2 == 0:
        hap += i
    i += 1
print(hap)
```

```python
i = 1
hap = 0
while i <= 50:
#     print(i, end=" ")
    if i % 3 == 0 or i % 2 == 0:
        print(f"hap:{hap}, i:{i}")
        hap += i
    i += 1
print(hap)
```

### while의 무한반복
* while 뒤의 조건잎True일 경우 무한 반복
* while 문 내에서 if 조건문으로 break 걸어줘야 멈춤

```python
# 무한반복
while True:
    print(i, "조건반복", end=" ")
    if i>= 100:
        break
    i += 1
```

* 무한 반복이 되는 조건 : True
* Ture가 되는 경우: 숫자 1, 정수, 실수, 문자열, list, tuple, dict, True
* False가 되는 경우: 숫자 0, 0.0, "", '', [], (), {}(전부 비어있음), False

```python
i=1
while [1,2]:
    print(i, "조건반복", end=" ")
    if i>= 100:
        break
    i += 1
```

# 중첩 반복문
* 반복문을 2개 이상 겹쳐서 사용
* 바늘 시계의 원리와 같음
* 안쪽 반복문이 끝난 후 바깥쪽 반복문이 실행
* 초침-> 분침-> 시침

```python
* 구구단 만들기 - 2중 중첩 for 문
* 2단-9단
* 1-9
* 2*1=2
```

```python
for num1 in range(2,10):
#     print(num1)
    for num2 in range (1,10):
        print(f"{num1} X {num2} = ",num1 * num2)
```

* 반복문 3개로 시계 만들기
* 초 00-59
* 분 00-59
* 시 00-23
* import time
* time.sleep(0.2)

```python
# import time
# time.sleep(0.2)
# for hour in range(0,24):
#     print(hour)
#     for minute in range (0, 60):
#         if minute < 60 :
#             hour += 1
#         print(minute)
#         for sec in range (0, 60):
#             print(sec)
#             print(f"{hour}, {minute}, {sec} = ", hour, minute, sec)
```

```python
import time
for si in range(0, 24):
    for bun in range(0, 60):
        for cho in range(0,60):
            print(f"{si:02d}:{bun:02d}:{cho:02d}", end="\r")
            time.sleep(1)
```
