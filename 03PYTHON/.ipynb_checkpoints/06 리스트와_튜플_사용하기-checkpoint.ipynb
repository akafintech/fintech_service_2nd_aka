# 튜플 tuple
* 튜플은 리스트처럼 모든 자료형을 넣어 만들 수 있다
* 순서가 있는 자료형, 인덱스가 있다.
* 인덱싱, 슬라이싱 가능
* 자료의 추가 삭제가 안됨
* () 안에 자료를 넣어서 만든다.
* 튜플 형변환 함수는 tuple()
* 자료 1개만 넣어서 만들 수 있다. (1,) 형식어야 한다.
* 튜플 생성시 자료가 2개 이상인 경우 ()는 생략 가능.

```python
t = (1,2,3)
t
```

```python
t[0]
```

```python
t[::-1]
```

```python
t2 = (4,5,6)
```

```python
t+t2
```

```python
tuple([1,2,3,4,5])
```

# 시퀀스 객체(자료형)
* 문자열, 리스트, 튜플 같은 순서가 있는 자료형 : 시퀀스 자료형
* 순서가 있고, 인덱스 번호가 부여되며 인덱싱, 슬라이싱이 가능
* 반복문에 시퀀스 자료형을 그대로 넣어서 반복시키는 것이 가능

* 특정값이 있는지 확인 in, not in

```python
t2
```

```python
4 in t2
```

```python
7 in t2
```

```python
c = "시퀀스 자료형은 in, not in으로 자료가 내부에 있는지 확인 가능"
```

```python
"자료" in c
```

```python
"확인"in c
```

```python
list_01=[1,2,3,4,5]
3 in list_01
```

* 시퀀스 객체는 + 연산자를 이용해서 서로 연결이 가능

```python
"시퀀스" + "자료형"
```

```python
[1,2,3,4] + [5,6,7,8]
```

```python
(1,2,3,4)+(5,6,7,8)
```

* 시퀀스 객체는 * 로 반복 가능

```python
"="*50
```

```python
[1,2,3]*3
```

```python
(4,5,6)*3
```

```python
for i in "문자열은 반복문에서 사용가능":
    print(i)
```

```python
for i in [1,2,3,4]:
    print(i)
```

```python
t1 = (1,2,3,4,5)
for i in t1:
    print(i)
```

* 리스트, 튜플은 sum(), max(), min()과 같은 함수로 연산도 가능

```python
sum(t1)
```

```python
max(t1)
```

```python
min(t1)
```
