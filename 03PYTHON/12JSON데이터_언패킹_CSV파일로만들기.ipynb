* 국민은행 앱 사욛자 리뷰 데이터 JSON 파일을 읽어와서 csv 파일로 변환하기

```python
import json
import pandas as pd
```

```python
with open("./data/com_kbstar_kbbank.json", 'r', encoding="utf-8") as file :
    data = file.read()
print(data[:20])
```

```python
data[:100]
```

```python
reviews = json.loads(data)
len(reviews)
```

```python
reviews[0]
```

```python
users=[]
points=[]
rev=[]
year_m=[]
recomd=[]

for i in reviews :
    users.append(i['사용자'])
    points.append(i['별점'])
    rev.append(i['리뷰'])
    year_m.append(i['날짜'])
    recomd.append(i['개발자 응답'])
    
review_info = dict(사용자=users, 별점=points, 리뷰=rev, 날짜=year_m, 개발자응답=recomd)
review_info
```

```python
df = pd.DataFrame(review_info)
df
```

```python
result={}
for key in reviews[0] :
    value_list = []
    for item in reviews[:2]:
        value_list.append(item[key])
    print(key, value_list)
    result[key] = value_list
```

* 풀이

```python
result = {}
for key in reviews[0]:
    temp = []
    for review in reviews:
        temp.append(review[key])
    result[key] = temp
len(result['별점'])
```

```python
df = pd.DataFrame(result)
df
```

* 별점은 숫자로 변환해서 넣기, 나머지 특문은 제거

```python
result = {}
for key in reviews[0]:
    temp = []
    for review in reviews:
        if key == '별점':
            temp.append(int(review[key]))
        else :
            temp.append(review[key].apepnd)
        temp.append(review[key])
    result[key] = temp
len(result['별점'])
```

```python
result2 = {}
for review in reviews[:2]:
    for key, value in review.items():
        result2.setdefault(key, []).append(value)
result2_df=pd.DataFrame(result2)
result2_df
```
