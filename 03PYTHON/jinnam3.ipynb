```python
import json
import pandas as pd
```

```python
with open("./data/hotels_korea_reviews_text.json", "r", encoding="utf-8") as file1 :
    temp = file1.read()
    data = json.loads(temp)
print(len(data))
```

```python
# 진남이 코드

hotel_infomations = []
reviews = []
for inner_data in data:
    body = inner_data['data']['body']
    hotel_id = body["pdpHeader"]['hotelId']
    
    propertyDescription = body["propertyDescription"]
    hotelName = propertyDescription['name']
    hotelGrade = propertyDescription['starRatingTitle']
    hotelRating = propertyDescription['starRating']
    region = propertyDescription['address'].get("region")
    locality = propertyDescription['address'].get("locality")

    ratingAspects = body["reviewContent"]['overall']["ratingAspects"]
    cleanliness = ratingAspects.get('cleanliness')
    service = ratingAspects.get('service')
    comfort = ratingAspects.get("comfort")
    condition = ratingAspects.get('condition')
    neighbourhood = ratingAspects.get('neighbourhood')
    
    hotel_information = {}
    hotel_information["hotel_id"] = hotel_id
    hotel_information["hotel_name"] = hotelName
    hotel_information["hotel_grade"] = hotelGrade
    hotel_information["n_reviews"] = hotelGrade
    hotel_information["avg_rating"] = hotelRating
    hotel_information["cleanliness"] = cleanliness
    hotel_information["service"] = service
    hotel_information["comfort"] = comfort
    hotel_information["condition"] = condition
    hotel_information["neighbourhood"] = neighbourhood
    hotel_information["region"] = region
    hotel_information["locality"] = locality
    hotel_infomations.append(hotel_information)

    

    groups = body["reviewContent"]["reviews"]["hermes"].get('groups')
    
    if groups:
        for group in groups:
            items = group['items']
            for item in items:
                review = {}
                review["hotel_id"] = hotel_id
                review["itineraryId"] = item.get('itineraryId')
                review["tripType"] = item['tripType']
                review["reviewDate"] = item['reviewDate']
                review["rating"] = item['rating']
                review["desceription"] = item['description']
                reviews.append(review)
```

```python
df_hotel_info = pd.DataFrame(hotel_infomations)
df_hotel_info
```
